```julia
using SparseArrays, Plots

Nvec = 2 .^ (3:12)
errors = Float64[]

plot(xaxis=:log, yaxis=:log, xlabel="N", ylabel="Error",
     title="Convergence of 4th-order finite differences", legend=false)

for N in Nvec
    h = 2π / N
    x = -π .+ (1:N) * h
    u = exp.(sin.(x))
    uprime = cos.(x) .* u

    # Construct sparse 4th-order periodic differentiation matrix
    e = ones(N)
    D = spzeros(N, N)
    for i in 1:N
        D[i, mod1(i+1, N)] +=  2/3
        D[i, mod1(i+2, N)] += -1/12
        D[i, mod1(i-1, N)] += -2/3
        D[i, mod1(i-2, N)] +=  1/12
    end
    D = (D - D') / h

    # Compute error
    err = norm(D * u - uprime, Inf)
    push!(errors, err)
    scatter!([N], [err], markersize=4, label="")
end

# Plot reference line for N^{-4} convergence
plot!(Nvec, Nvec .^ -4, linetype=:dash, label="N^{-4}")
annotate!(100, 5e-8, text("N^{-4}", :left, 10))
```
